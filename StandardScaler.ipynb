# Standard Scaling or Standardizing the data or z_score

```python
import pandas as pd 
from sklearn.preprocessing import StandardScaler
```

```python
df ={
    "age": [25,30,35,45,50],
    "height":[165,170,180,185,190],
    "weight":[55,60,65,70,75]
}

# convert this data to pandas dataframe
df = pd.DataFrame(df)
df.head()
```

```python
# import the scaler 
scalar = StandardScaler()
scaled_df = scalar.fit_transform(df)
scaled_df
```

```python
# convet this dataframe into Pandas dataframe 
scaled_df = pd.DataFrame(scaled_df,columns=df.columns)
scaled_df.head()
```
